```python
import ee
import geemap
import os

try:
    ee.Initialize()
    print("GEE Initialized.")
except Exception as e:
    print(f"Error GEE: {e}")
```

```python
AOI = ee.Geometry.Rectangle([-66.35, -17.50, -65.90, -17.20])

# 2. Crear el objeto Mapa
Map = geemap.Map()

# 3. Centrar el mapa y añadir la capa
Map.centerObject(AOI, 11)
Map.addLayer(AOI, {'color': 'red'}, 'Mi Área de Interés')

# 4. Mostrar el mapa
Map
```
